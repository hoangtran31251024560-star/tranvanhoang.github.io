```python
import folium
import pandas as pd
from geopy.geocoders import Nominatim
from geopy import distance

geolocator = Nominatim(user_agent='geo app', timeout=10)
addresses = [
    "Dinh Độc Lập, Quận 1, TP.HCM",
    "Nhà thờ Đức Bà, Quận 1, TP.HCM",
    "Chợ Bến Thành, Quận 1, TP.HCM",
    "Tòa nhà Bitexco, Quận 1, TP.HCM",
    "Landmark 81, Bình Thạnh, TP.HCM",
    "Nhà hát Thành phố, Quận 1, TP.HCM",
    "Bảo tàng Chứng tích Chiến tranh, Quận 3, TP.HCM",
    "Thảo Cầm Viên, Quận 1, TP.HCM",
    "Phố đi bộ Nguyễn Huệ, Quận 1, TP.HCM",
    "Sân bay Tân Sơn Nhất, Tân Bình, TP.HCM",

]

center_location = geolocator.geocode("UEH University, Ho Chi Minh City, Vietnam")
center_coords = (center_location.latitude, center_location.longitude)
map = folium.Map(location=center_coords, zoom_start=15)
folium.Marker(location=center_coords, popup='<b>UEH University, Ho Chi Minh City, Vietnam<b>', icon=folium.Icon(color='blue', icon='university', prefix='fa')).add_to(map)

for addr in addresses:
  location = geolocator.geocode(addr)
  if location:
    l_coords = (location.latitude, location.longitude)
    dist = distance.distance(center_coords, l_coords).km
    folium.Marker(l_coords, popup=f'{addr}: {dist:.2f}km').add_to(map)
    folium.PolyLine([center_coords, l_coords], color='red', weight=1).add_to(map)
map
```

```python
import geopandas as gpd
import pandas as pd
import folium

url = "https://gist.githubusercontent.com/hoanganh25991/00e8f0162e41ac31ecd7/raw/district-boundary-hcm-city.geojson"

gdf = gpd.read_file(url)


gdf = gdf[['name', 'geometry']]

data = pd.DataFrame({
    'district': ["District 1", "District 3", "District 4", "District 5"],
    'orders': [500, 376, 905, 98]
})

gdf = gdf.merge(data, left_on='name', right_on='district')
m = folium.Map(location=[10.76,106.68], zoom_start=11)

folium.Choropleth(
    geo_data=gdf.to_json(),
    data=gdf,
    columns=['district', 'orders'],
    key_on='feature.properties.name',
    fill_color='YlOrRd',
    fill_opacity=0.8,
    line_opacity=0.2,
    legend_name='Số lượng hàng'
).add_to(m)

m
```

```python
import pandas as pd
import folium
from geopy.geocoders import Nominatim
from geopy.distance import geodesic
geolocator=Nominatim(user_agent='service_area_app')
center_location="Ben Thanh Market, Ho Chi Minh City, Vietnam"
location=geolocator.geocode(center_location)
center_coords=(location.latitude,location.longitude)
map1=folium.Map(location=center_coords,zoom_start=15)

Customers=pd.DataFrame({
    'customers':['hk1','hk2','hk3','hk4','hk5','hk6'],
    'latitude':[10.7750, 10.7820, 10.7600, 10.7300, 10.8100,
        10.8500],
    'longitude':[106.7000, 106.6900, 106.7200, 106.7100, 106.6800,
        106.7600]

})
for i in range(len(Customers)):
  customers_coords=(
      Customers.loc[i,'latitude'],
      Customers.loc[i,'longitude']
  )
distance=geodesic(center_coords,customers_coords).km
Customers['distance']=distance
def able_zone(d):
  if d<=3:
    return '0-3 km'
  elif d<=5 :
    return '3-5 km'
  elif d<=10:
    return '5-10 km'
  else:
    return '> 10 km'
Customers['zone']=Customers['distance'].apply(able_zone)
sumary=Customers['zone'].value_counts()
print(sumary,'hanh khach')

folium.Marker(location=center_coords,popup='<b>Trung tam van chuyen<b>', icon=folium.Icon(color='red',icon='delivery',prefix='fa')).add_to(map1)

for i in range(len(Customers)):
  folium.Marker(location=[
  Customers.loc[i,'latitude'],
  Customers.loc[i,'longitude'],],
  color='red'


    ).add_to(map1)

folium.Circle(location=center_coords,color='red',radius=3000,fill=True,popup='vung 3 km ').add_to(map1)
folium.Circle(location=center_coords,color='green',radius=5000,fill=True,popup='vung 5 km ').add_to(map1)
folium.Circle(location=center_coords,color='blue',radius=10000,fill=True,popup='vung 10 km ').add_to(map1)



map1



```
