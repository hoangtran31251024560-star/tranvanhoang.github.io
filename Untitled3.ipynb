```python
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import numpy as np
distance=ctrl.Antecedent(np.arange(0,51,1),'distance')
traffic=ctrl.Antecedent(np.arange(0,101,1),'traffic')
demand=ctrl.Antecedent(np.arange(0,101,1),'demand')
weather=ctrl.Antecedent(np.arange(0,11,1),'weather')
rating=ctrl.Antecedent(np.arange(1,5.1,0.1),'rating')
punctuality=ctrl.Antecedent(np.arange(0,101,1),'punctuality')
price=ctrl.Consequent(np.arange(0,101,1), 'price')
bonus=ctrl.Consequent(np.arange(0,101,1), 'bonus')
distance['Short'] = fuzz.trimf(distance.universe, [0, 0, 3])
distance['Medium'] = fuzz.trimf(distance.universe, [2, 5, 8])
distance['Long'] = fuzz.trimf(distance.universe, [6, 13, 20])
distance['Very Long'] = fuzz.trimf(distance.universe, [15, 50, 50])

traffic['Low'] = fuzz.trimf(traffic.universe, [0, 0, 30])
traffic['Medium'] = fuzz.trimf(traffic.universe, [20, 45, 70])
traffic['High'] = fuzz.trimf(traffic.universe, [60, 100, 100])
demand['Low'] = fuzz.trimf(demand.universe, [0, 0, 30])
demand['Medium'] = fuzz.trimf(demand.universe, [20, 45, 70])
demand['High'] = fuzz.trimf(demand.universe, [60, 100, 100])
rating['Poor'] = fuzz.trimf(rating.universe, [1, 1, 2.5])
rating['Average'] = fuzz.trimf(rating.universe, [2, 3, 4])
rating['Good'] = fuzz.trimf(rating.universe, [3.5, 5, 5])
punctuality['Late'] = fuzz.trimf(punctuality.universe, [0, 0, 50])
punctuality['OnTime'] = fuzz.trimf(punctuality.universe, [40, 60, 80])
punctuality['Early'] = fuzz.trimf(punctuality.universe, [70, 100, 100])
weather['Good'] = fuzz.trimf(weather.universe, [0, 0, 5])
weather['Moderate'] = fuzz.trimf(weather.universe, [0, 5, 10])
weather['Bad'] = fuzz.trimf(weather.universe, [5, 10, 10])
price['Low']=fuzz.trimf(price.universe,[0,0,25])
price['Medium']=fuzz.trimf(price.universe,[0,25,50])
price['High']=fuzz.trimf(price.universe,[25,50,50])
price['Very High']=fuzz.trimf(price.universe,[50,100,100])
bonus['Low']=fuzz.trimf(bonus.universe,[0,0,25])
bonus['Medium']=fuzz.trimf(bonus.universe,[0,25,50])
bonus['High']=fuzz.trimf(bonus.universe,[25,50,50])
bonus['Very High']=fuzz.trimf(bonus.universe,[50,100,100])

rule1=ctrl.Rule(distance['Short']&traffic['Low']&demand['Low'],price['Low'])
rule2=ctrl.Rule(distance['Short']&traffic['Medium']&demand['High'],price['Medium'])
rule3=ctrl.Rule(distance['Medium']&traffic['High']&demand['High'],price['High'])
rule4=ctrl.Rule(distance['Long']&traffic['Medium']&weather['Good'],price['Medium'])
rule5=ctrl.Rule(distance['Long']&traffic['High']&weather['Bad'],price['Very High'])
rule6=ctrl.Rule(distance['Very Long']&traffic['High']&demand['High'],price['Very High'])
rule7 = ctrl.Rule(distance['Medium'] & traffic['Low'] & demand['Low'], price['Medium'])
rule8 = ctrl.Rule(distance['Short'] & traffic['High'] & weather['Bad'], price['High'])
rule9 = ctrl.Rule(distance['Very Long'] & weather['Bad'], price['Very High'])
rule10 = ctrl.Rule(distance['Medium'] & traffic['Medium'] & weather['Moderate'], price['Medium'])
rule11 = ctrl.Rule(rating['Good'] & punctuality['Early'], bonus['High'])
rule12 = ctrl.Rule(rating['Average'] & punctuality['OnTime'], bonus['Medium'])
rule13 = ctrl.Rule(rating['Poor'] & punctuality['Late'], bonus['Low'])

rule14 = ctrl.Rule(distance['Long'] & traffic['High'] & punctuality['OnTime'], bonus['High'])
rule15 = ctrl.Rule(distance['Medium'] & traffic['Medium'] & rating['Good'], bonus['Medium'])
rule16 = ctrl.Rule(rating['Poor'] & punctuality['Late'], bonus['Low'])
rule17 = ctrl.Rule(distance['Very Long'] & weather['Bad'] & rating['Good'], bonus['High'])
rule18 = ctrl.Rule(distance['Short'] & rating['Average'] & punctuality['OnTime'], bonus['Low'])
rule19 = ctrl.Rule(distance['Long'] & traffic['High'] & punctuality['Late'], bonus['Low'])
rule20 = ctrl.Rule(distance['Medium'] & weather['Moderate'] & rating['Good'], bonus['Medium'])




price_rules = [rule1, rule2, rule3, rule4, rule5, rule6, rule7, rule8, rule9, rule10]
price_ctrl = ctrl.ControlSystem(price_rules)
price_sim = ctrl.ControlSystemSimulation(price_ctrl)
bonus_rules = [rule11, rule12, rule13, rule14, rule15, rule16, rule17, rule18, rule19, rule20]
bonus_ctrl = ctrl.ControlSystem(bonus_rules)
bonus_sim = ctrl.ControlSystemSimulation(bonus_ctrl)


price_sim.input['distance'] = 25
price_sim.input['traffic'] = 80
price_sim.input['weather'] = 9
price_sim.input['demand'] = 50


bonus_sim.input['rating'] = 4.5
bonus_sim.input['punctuality'] = 90
bonus_sim.input['distance'] = 25
bonus_sim.input['traffic'] = 80
bonus_sim.input['weather'] = 9


price_sim.compute()
bonus_sim.compute()
print(f"Kết quả Giá: {price_sim.output['price']:.2f}")
print(f"Kết quả Thưởng: {bonus_sim.output['bonus']:.2f}")
```

```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

rating = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'rating')
volume = ctrl.Antecedent(np.arange(0, 101, 1), 'volume')
profit = ctrl.Antecedent(np.arange(0, 101, 1), 'profit')
event = ctrl.Antecedent(np.arange(0, 11, 1), 'event')
discount = ctrl.Antecedent(np.arange(0, 101, 1), 'discount')
chiet_khau = ctrl.Consequent(np.arange(0, 71, 1), 'chiet_khau')

rating['Low'] = fuzz.trimf(rating.universe, [0, 0, 4.0])
rating['Medium'] = fuzz.trimf(rating.universe, [4.0, 4.25, 4.5])
rating['High'] = fuzz.trimf(rating.universe, [4.5, 5, 5])

volume['Low'] = fuzz.trimf(volume.universe, [0, 0, 30])
volume['Medium'] = fuzz.trimf(volume.universe, [20, 50, 80])
volume['High'] = fuzz.trimf(volume.universe, [70, 100, 100])

profit['Low'] = fuzz.trimf(profit.universe, [0, 0, 30])
profit['Medium'] = fuzz.trimf(profit.universe, [20, 50, 80])
profit['High'] = fuzz.trimf(profit.universe, [70, 100, 100])

event['Low'] = fuzz.trimf(event.universe, [0, 0, 3])
event['Medium'] = fuzz.trimf(event.universe, [2, 5, 8])
event['High'] = fuzz.trimf(event.universe, [7, 10, 10])

discount['Low'] = fuzz.trimf(discount.universe, [0, 0, 30])
discount['Medium'] = fuzz.trimf(discount.universe, [20, 50, 80])
discount['High'] = fuzz.trimf(discount.universe, [70, 100, 100])

chiet_khau['very Low'] = fuzz.trimf(chiet_khau.universe, [0, 0, 5])
chiet_khau['Low'] = fuzz.trimf(chiet_khau.universe, [5, 7.5, 10])
chiet_khau['Medium'] = fuzz.trimf(chiet_khau.universe, [10, 15, 20])
chiet_khau['High'] = fuzz.trimf(chiet_khau.universe, [20, 30, 40])
chiet_khau['Very High'] = fuzz.trimf(chiet_khau.universe, [40, 55, 70])

rule1 = ctrl.Rule(rating['High'] & volume['High'] & profit['High'], chiet_khau['very Low'])
rule2 = ctrl.Rule(rating['Low'] & volume['Low'] & profit['High'], chiet_khau['Very High'])
rule3 = ctrl.Rule(event['High'] & discount['High'], chiet_khau['Very High'])
rule4 = ctrl.Rule(rating['Medium'] & volume['Medium'] & profit['Medium'], chiet_khau['Medium'])
rule5 = ctrl.Rule(discount['Low'] & volume['High'] & profit['Low'], chiet_khau['very Low'])
rule6 = ctrl.Rule(rating['Low'] & event['Low'], chiet_khau['Medium'])
rule7 = ctrl.Rule(volume['Low'] & profit['Low'], chiet_khau['Very High'])

chiet_khau_ctrl = ctrl.ControlSystem([rule1, rule2, rule3, rule4, rule5, rule6, rule7])
tinh_chiet_khau = ctrl.ControlSystemSimulation(chiet_khau_ctrl)

tinh_chiet_khau.input['rating'] = 4.3
tinh_chiet_khau.input['volume'] = 50
tinh_chiet_khau.input['profit'] = 15
tinh_chiet_khau.input['event'] = 10
tinh_chiet_khau.input['discount'] = 90

tinh_chiet_khau.compute()
print(tinh_chiet_khau.output['chiet_khau'])
```

```python
import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl

nhu_cau = ctrl.Antecedent(np.arange(0, 101, 1), 'nhu_cau')
ap_luc = ctrl.Antecedent(np.arange(0, 101, 1), 'ap_luc')
uy_tin = ctrl.Antecedent(np.arange(0, 5.1, 0.1), 'uy_tin')
loi_nhuan = ctrl.Antecedent(np.arange(0, 101, 1), 'loi_nhuan')
mua_vu = ctrl.Antecedent(np.arange(0, 11, 1), 'mua_vu')
giam_gia = ctrl.Consequent(np.arange(0, 71, 1), 'giam_gia')

nhu_cau['Thấp'] = fuzz.trimf(nhu_cau.universe, [0, 0, 50])
nhu_cau['Trung bình'] = fuzz.trimf(nhu_cau.universe, [20, 50, 80])
nhu_cau['Cao'] = fuzz.trimf(nhu_cau.universe, [50, 100, 100])

ap_luc['Thấp'] = fuzz.trimf(ap_luc.universe, [0, 0, 50])
ap_luc['Trung bình'] = fuzz.trimf(ap_luc.universe, [20, 50, 80])
ap_luc['Cao'] = fuzz.trimf(ap_luc.universe, [50, 100, 100])

uy_tin['Thấp'] = fuzz.trimf(uy_tin.universe, [0, 0, 4.0])
uy_tin['Trung bình'] = fuzz.trimf(uy_tin.universe, [4.0, 4.25, 4.5])
uy_tin['Cao'] = fuzz.trimf(uy_tin.universe, [4.5, 5.0, 5.0])

loi_nhuan['Thấp'] = fuzz.trimf(loi_nhuan.universe, [0, 0, 50])
loi_nhuan['Trung bình'] = fuzz.trimf(loi_nhuan.universe, [20, 50, 80])
loi_nhuan['Cao'] = fuzz.trimf(loi_nhuan.universe, [50, 100, 100])

mua_vu['Không có'] = fuzz.trimf(mua_vu.universe, [0, 0, 4])
mua_vu['Trung bình'] = fuzz.trimf(mua_vu.universe, [3, 5, 7])
mua_vu['Cao'] = fuzz.trimf(mua_vu.universe, [6, 10, 10])

giam_gia['Rất thấp'] = fuzz.trimf(giam_gia.universe, [0, 2.5, 5])
giam_gia['Thấp'] = fuzz.trimf(giam_gia.universe, [5, 7.5, 10])
giam_gia['Trung bình'] = fuzz.trimf(giam_gia.universe, [10, 15, 20])
giam_gia['Cao'] = fuzz.trimf(giam_gia.universe, [20, 30, 40])
giam_gia['Rất cao'] = fuzz.trimf(giam_gia.universe, [40, 55, 70])

rule1 = ctrl.Rule(nhu_cau['Cao'] & ap_luc['Thấp'] & loi_nhuan['Thấp'], giam_gia['Rất thấp'])
rule2 = ctrl.Rule(nhu_cau['Thấp'] & ap_luc['Cao'] & loi_nhuan['Cao'], giam_gia['Cao'])
rule3 = ctrl.Rule(uy_tin['Cao'] & loi_nhuan['Trung bình'] & mua_vu['Cao'], giam_gia['Trung bình'])
rule4 = ctrl.Rule(ap_luc['Cao'] & mua_vu['Cao'] & loi_nhuan['Cao'], giam_gia['Rất cao'])
rule5 = ctrl.Rule(uy_tin['Thấp'] & nhu_cau['Trung bình'] & loi_nhuan['Thấp'], giam_gia['Trung bình'])
rule6 = ctrl.Rule(nhu_cau['Cao'] & mua_vu['Không có'] & ap_luc['Thấp'], giam_gia['Rất thấp'])
rule7 = ctrl.Rule(loi_nhuan['Cao'] & ap_luc['Trung bình'] & mua_vu['Trung bình'], giam_gia['Trung bình'])

giam_gia_ctrl = ctrl.ControlSystem([rule1, rule2, rule3, rule4, rule5, rule6, rule7])
he_thong = ctrl.ControlSystemSimulation(giam_gia_ctrl)

he_thong.input['nhu_cau'] = 90
he_thong.input['ap_luc'] = 50
he_thong.input['uy_tin'] = 4.2
he_thong.input['loi_nhuan'] = 85
he_thong.input['mua_vu'] = 10

he_thong.compute()
print(he_thong.output['giam_gia'])
```
